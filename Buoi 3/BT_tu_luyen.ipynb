Bài 1: List – Quản lý điểm

- Nhập n điểm sinh viên vào một list.

- Yêu cầu:

    - Điểm phải từ 0 đến 10, nhập sai thì nhập lại.
    - Tính điểm trung bình.
    - Tìm điểm lớn nhất, nhỏ nhất mà không dùng max(), min().
    - Tạo danh sách các điểm lớn hơn điểm trung bình.
    - Kiểm tra có sinh viên đạt điểm 10 hay không.

```python
# Nhập n điểm sinh viên vào một list
n = int(input("Nhập số lượng sinh viên n: "))
danh_sach_diem = [] #list rỗng chứa danh sách điểm

for i in range(n):
  while True:
    diem = float(input(f"Nhập điểm cho sinh viên thứ {i+1}: "))
    # Điểm phải từ 0 đến 10, nhập sai thì nhập lại
    if 0 <= diem <= 10:
      danh_sach_diem.append(diem) # điểm được nhập đưa vào list
      break
    else:
      print("Điểm không hợp lệ! Vui lòng nhập lại trong khoảng từ 0 đến 10.")

print("\n--- KẾT QUẢ ---")
print("Danh sách điểm:", danh_sach_diem)

# Tính điểm trung bình[cite: 5]
diem_tb = sum(danh_sach_diem) / n
print(f"- Điểm trung bình: {diem_tb:.2f}") # {diem_tb:.2f}: Định dạng hiển thị kết quả lấy 2 chữ số sau dấu phẩy.

# Tìm điểm lớn nhất, nhỏ nhất mà không dùng max(), min()[cite: 5]
diem_max = danh_sach_diem[0] # Giả định phần tử đầu tiên trong danh sách vừa là điểm lớn nhất, vừa là điểm nhỏ nhất ban đầu.
diem_min = danh_sach_diem[0]

for diem in danh_sach_diem:
  if diem > diem_max:
    diem_max = diem
  if diem < diem_min:
    diem_min = diem

print(f"- Điểm lớn nhất: {diem_max}")
print(f"- Điểm nhỏ nhất: {diem_min}")

# Tạo danh sách các điểm lớn hơn điểm trung bình[cite: 5]
diem_lon_hon_tb = [diem for diem in danh_sach_diem
                   if diem > diem_tb]
print(f"- Các điểm lớn hơn điểm trung bình: {diem_lon_hon_tb}")

# Kiểm tra có sinh viên đạt điểm 10 hay không[cite: 5]
if 10 in danh_sach_diem:
  print("- Có sinh viên đạt điểm 10.")
else:
  print("- Không có sinh viên nào đạt điểm 10.")
```

Bài 3: Set – Quản lý câu lạc bộ

- Có ba set lưu sinh viên tham gia:

    python_club
    ai_club
    english_club

- Yêu cầu:

    - Tìm sinh viên tham gia ít nhất một câu lạc bộ.
    - Tìm sinh viên tham gia cả ba câu lạc bộ.
    - Tìm sinh viên chỉ tham gia Python.
    - Tìm sinh viên tham gia đúng một câu lạc bộ.
    - Tìm sinh viên tham gia ít nhất hai câu lạc bộ.
    - Nhập mã sinh viên và in các câu lạc bộ sinh viên đó tham gia.

```python
# Ba set lưu sinh viên tham gia các câu lạc bộ (Dữ liệu mẫu)[cite: 5]
python_club = {"SV001", "SV002", "SV003", "SV004"}
ai_club = {"SV003", "SV004", "SV005", "SV006"}
english_club = {"SV001", "SV004", "SV007", "SV008"}

# 1. Tìm sinh viên tham gia ít nhất một câu lạc bộ (Phép hợp - union)[cite: 1, 5]
sv_it_nhat_mot = python_club.union(ai_club, english_club)
print("1. Sinh viên tham gia ít nhất một CLB:", sv_it_nhat_mot)

# 2. Tìm sinh viên tham gia cả ba câu lạc bộ (Phép giao - intersection)[cite: 1, 5]
sv_ca_ba = python_club.intersection(ai_club, english_club)
print("2. Sinh viên tham gia cả 3 CLB:", sv_ca_ba)

# 3. Tìm sinh viên chỉ tham gia Python (difference)[cite: 1, 5]
sv_chi_python = python_club.difference(ai_club, english_club)
print("3. Sinh viên chỉ tham gia Python:", sv_chi_python)

# 4. Tìm sinh viên tham gia đúng một câu lạc bộ (symmetric_difference / toán học tập)[cite: 1, 5]
sv_dung_mot_clb = (
    (python_club.symmetric_difference(ai_club)).symmetric_difference(
        english_club
    )
).difference(sv_ca_ba)
print("4. Sinh viên tham gia đúng một CLB:", sv_dung_mot_clb)

# 5. Tìm sinh viên tham gia ít nhất hai câu lạc bộ[cite: 5]
sv_it_nhat_hai = (
    (python_club.intersection(ai_club))
    .union(python_club.intersection(english_club))
    .union(ai_club.intersection(english_club))
)
print("5. Sinh viên tham gia ít nhất hai CLB:", sv_it_nhat_hai)

# 6. Nhập mã sinh viên và in các câu lạc bộ sinh viên đó tham gia[cite: 5]
ma_sv = input(
    "\nNhập mã sinh viên cần kiểm tra (Ví dụ: SV001): "
).strip()
clb_tham_gia = []

if ma_sv in python_club:
  clb_tham_gia.append("Python Club")
if ma_sv in ai_club:
  clb_tham_gia.append("AI Club")
if ma_sv in english_club:
  clb_tham_gia.append("English Club")

if clb_tham_gia:
  print(f"Sinh viên {ma_sv} tham gia các CLB: {', '.join(clb_tham_gia)}")
else:
  print(f"Sinh viên {ma_sv} không tham gia câu lạc bộ nào.")
```
